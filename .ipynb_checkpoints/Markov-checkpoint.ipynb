```python
# Importing dependencies
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import defaultdict, Counter
import random
import math
import urllib.request
import gzip
import zipfile
from pathlib import Path

# Setting random seed for output reproducability
random.seed(42)
```

```python
# Setting up default paths
DATA_DIR = Path("data")
SEQ_FILE = DATA_DIR / "msnbc990928.seq"
SOURCE_URL = "https://archive.ics.uci.edu/static/public/133/msnbc+com+anonymous+web+data.zip"
```

```python
# This function checks if the dataset exists locally otherwise it downloads it for usage.
def ensure_msnbc_dataset(url = SOURCE_URL, file_path = SEQ_FILE):
    if file_path.exists():
        return file_path
    file_path.parent.mkdir(parents=True, exist_ok=True)
    zip_path = file_path.parent / "msnbc.zip"
    print(f"Local dataset not found at location '{file_path}'. Downloading...")
    req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0'})

    with urllib.request.urlopen(req) as response, open(zip_path, 'wb') as out_file:
        out_file.write(response.read())

    with zipfile.ZipFile(zip_path, 'r') as z:
        # Extract the .seq file directly into the data folder
        gz_file = next(f for f in z.namelist() if f.endswith('seq.gz') or f.endswith('.seq'))
        gz_data = z.read(gz_file)

    if gz_file.endswith('.gz'):
        raw_seq = gzip.decompress(gz_data)
    else:
        raw_seq = gz_data

    with open(file_path, 'wb') as f:
        f.write(raw_seq)

    if zip_path.exists():
        zip_path.unlink()

    print(f"Dataset successfully downloaded to - '{file_path}'.")
    return file_path
```

```python
def load_msnbc_sequences(file_path=SEQ_FILE):
    """Reading and parsing the MSNBC sequence dataset line-by-line."""
    sequences = []
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line or line.startswith('%'):
                continue
            seq = [int(x) for x in line.split() if x.isdigit()]
            if seq:
                sequences.append(seq)
    return sequences
```

```python
ensure_msnbc_dataset()
data = load_msnbc_sequences()
random.shuffle(data)
split_idx = int(len(data) * 0.8)
train_data = data[:split_idx]
test_data = data[split_idx:]
print(f"Total sequences: {len(data)}, Train: {len(train_data)}, Test: {len(test_data)}")
```

```python
def build_markov_model(
    sequences, 
    k, 
    tau = 0
):
    """
    Builds a k-order Markov model transition table from sequence data.
    Optionally prunes history contexts observed fewer than `tau` times.
    """
    transitions = defaultdict(Counter)

    for seq in sequences:
        if len(seq) <= k:
            continue
        for i in range(len(seq) - k):
            history = tuple(seq[i:(i + k)])
            next_state = seq[i + k]
            transitions[history][next_state] += 1

    # Prune low-frequency history contexts
    if tau > 0:
        return {h: counts for h, counts in transitions.items() if sum(counts.values()) >= tau}

    return dict(transitions)
```

```python
def predict_and_evaluate(
    model, 
    test_sequences, 
    k, 
    alpha = 1.0, 
    num_states = 17
):
    """
    Evaluates a k-order model using Laplace (Add-alpha) smoothing.
    Returns (Average Log-Likelihood, Top-1 Accuracy).
    """
    total_log_likelihood = 0.0
    correct_predictions = 0
    total_predictions = 0

    for seq in test_sequences:
        if len(seq) <= k:
            continue

        for i in range(len(seq) - k):
            history = tuple(seq[i:i + k])
            actual_next = seq[i + k]

            counts = model.get(history, Counter())
            total_observations = sum(counts.values())
            denominator = total_observations + (alpha * num_states)

            # Top-1 prediction (argmax over smoothed probabilities)
            if total_observations == 0:
            # Example: fall back to uniform/random or global unigram
                best_state = random.randint(1, num_states)
            # Under uniform Laplace smoothing, argmax P(s|h) equals argmax count(h, s)
            best_state = max(
                range(1, num_states + 1),
                key=lambda s: counts.get(s, 0)
            )

            if best_state == actual_next:
                correct_predictions += 1

            # Compute log-likelihood for observed state
            actual_prob = (counts.get(actual_next, 0) + alpha) / denominator
            total_log_likelihood += math.log(actual_prob)
            total_predictions += 1

    if total_predictions == 0:
        return 0.0, 0.0

    avg_log_likelihood = total_log_likelihood / total_predictions
    accuracy = correct_predictions / total_predictions

    return avg_log_likelihood, accuracy
```

```python
K_VALUES = [1, 2, 3, 4, 5, 6]
TAU_VALUES = [0, 5, 10]
NUM_STATES = 17
ALPHA = 1.0

results = []

print(f"Running grid search over orders k={K_VALUES} and pruning thresholds tau={TAU_VALUES}...")

for k in K_VALUES:
    for tau in TAU_VALUES:
        # Build Markov model on training sequences
        model = build_markov_model(train_data, k=k, tau=tau)
        
        # Evaluate model on held-out test set
        avg_ll, acc = predict_and_evaluate(
            model=model,
            test_sequences=test_data,
            k=k,
            alpha=ALPHA,
            num_states=NUM_STATES
        )
        
        results.append({
            'Order (k)': k,
            'Threshold (tau)': tau,
            'Active States': len(model),
            'Max Theoretical States': NUM_STATES ** k,
            'Accuracy': acc,
            'Avg Log-Likelihood': avg_ll
        })

# Create results DataFrame
df_results = pd.DataFrame(results)

# --- File Exports for Paper & Excel ---
TABLES_DIR = Path("tables")
TABLES_DIR.mkdir(exist_ok=True)

# 1. Export silent LaTeX snippet (for \input{tables/results.tex} in paper)
df_results.to_latex(
    TABLES_DIR / "results.tex",
    index=False,
    float_format="%.4f",
    caption="Predictive performance and state-space complexity across Markov orders (k) and pruning thresholds (\\tau).",
    label="tab:markov_results"
)

# 2. Export Excel / CSV copies for offline inspection
df_results.to_csv(TABLES_DIR / "results.csv", index=False)
# df_results.to_excel(TABLES_DIR / "results.xlsx", index=False) # Requires openpyxl

print("Experiment execution complete. Results saved to 'tables/'.")

# --- In-Notebook HTML Table Rendering ---
# Displays formatted numbers, thousands separators, and highlights max values
(
    df_results.style
    .format({
        'Active States': '{:,}',
        'Max Theoretical States': '{:,}',
        'Accuracy': '{:.4f}',
        'Avg Log-Likelihood': '{:.4f}'
    })
    .highlight_max(subset=['Accuracy'], color='#d1e7dd')
    .highlight_max(subset=['Avg Log-Likelihood'], color='#d1e7dd')
    .set_caption("Table 1: Markov Chain Performance & State-Space Expansion")
    .hide(axis='index')
)
```

```python
# Force Jupyter to display inline plots as vector SVGs instead of raster PNGs
try:
    get_ipython().run_line_magic('config', "InlineBackend.figure_format = 'svg'")
except NameError:
    pass

# Setup figures directory
FIGURES_DIR = Path("figures")
FIGURES_DIR.mkdir(exist_ok=True)

# Academic styling preferences
sns.set_theme(style="whitegrid", font="sans-serif")
plt.rcParams.update({
    'font.size': 11,
    'axes.labelsize': 12,
    'axes.titlesize': 13,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'legend.fontsize': 10
})

CATEGORIES = [
    "frontpage", "news", "tech", "local", "opinion", "on-air", "misc",
    "weather", "health", "living", "business", "sports", "summary",
    "bbs", "travel", "msn-news", "msn-sports"
]

# ---------------------------------------------------------
# 1. State-Space Expansion Plot: Active States vs. Order (k)
# ---------------------------------------------------------
plt.figure(figsize=(8, 5))
ax = sns.lineplot(
    data=df_results, 
    x='Order (k)', 
    y='Active States', 
    hue='Threshold (tau)', 
    marker='o', 
    palette='crest',
    linewidth=2
)
plt.yscale('log')
plt.title('State-Space Context Growth vs. Markov Order ($k$)', pad=12)
plt.xlabel('Markov Order ($k$)')
plt.ylabel('Observed Contexts $|S^k_{obs}|$ (Log Scale)')
plt.xticks(K_VALUES)
plt.legend(title=r'Pruning $\tau$', frameon=True)

# Save SVG for paper
plt.savefig(FIGURES_DIR / "active_states_vs_k.svg", format='svg', bbox_inches='tight')
plt.show()


# ---------------------------------------------------------
# 2. Performance Metrics Plot: Accuracy & Log-Likelihood
# ---------------------------------------------------------
fig, ax_acc = plt.subplots(figsize=(8, 5))
df_unpruned = df_results[df_results['Threshold (tau)'] == 0]

color_acc = '#1f77b4'
color_ll = '#d62728'

# Primary axis: Accuracy
line1 = ax_acc.plot(
    df_unpruned['Order (k)'], 
    df_unpruned['Accuracy'], 
    color=color_acc, 
    marker='o', 
    linewidth=2, 
    label='Top-1 Accuracy'
)
ax_acc.set_xlabel('Markov Order ($k$)')
ax_acc.set_ylabel('Top-1 Accuracy', color=color_acc)
ax_acc.tick_params(axis='y', labelcolor=color_acc)
ax_acc.set_xticks(K_VALUES)

# Secondary axis: Avg Log-Likelihood
ax_ll = ax_acc.twinx()
line2 = ax_ll.plot(
    df_unpruned['Order (k)'], 
    df_unpruned['Avg Log-Likelihood'], 
    color=color_ll, 
    marker='s', 
    linestyle='--', 
    linewidth=2, 
    label='Avg Log-Likelihood'
)
ax_ll.set_ylabel('Average Log-Likelihood', color=color_ll)
ax_ll.tick_params(axis='y', labelcolor=color_ll)
ax_ll.grid(False)  # Turn off twin axis grid to prevent visual clutter

# Combined Legend
lines = line1 + line2
labels = [l.get_label() for l in lines]
ax_acc.legend(lines, labels, loc='lower right', frameon=True)

plt.title('Predictive Performance Metrics vs. Order ($k$) [$\\tau=0$]', pad=12)
fig.tight_layout()

# Save SVG for paper
plt.savefig(FIGURES_DIR / "performance_vs_k.svg", format='svg', bbox_inches='tight')
plt.show()


# ---------------------------------------------------------
# 3. Transition Matrix Heatmap (First-Order Model)
# ---------------------------------------------------------
first_order_model = build_markov_model(train_data, k=1, tau=0)
trans_matrix = np.zeros((NUM_STATES, NUM_STATES))

for i in range(1, NUM_STATES + 1):
    counts = first_order_model.get((i,), Counter())
    denom = sum(counts.values()) + (ALPHA * NUM_STATES)
    for j in range(1, NUM_STATES + 1):
        trans_matrix[i - 1, j - 1] = (counts.get(j, 0) + ALPHA) / denom

plt.figure(figsize=(9.5, 8))
sns.heatmap(
    trans_matrix, 
    annot=False, 
    cmap="YlGnBu", 
    xticklabels=CATEGORIES, 
    yticklabels=CATEGORIES,
    cbar_kws={'label': r'Transition Probability $P(S_{t+1} \mid S_t)$'}
)
plt.title('First-Order State Transition Probability Matrix (Laplace Smoothed)', pad=12)
plt.xlabel('Next Category ($S_{t+1}$)')
plt.ylabel('Current Category ($S_t$)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

# Save SVG for paper
plt.savefig(FIGURES_DIR / "transition_matrix_heatmap.svg", format='svg', bbox_inches='tight')
plt.show()
```
